<h2>LOAD DATA SET</h2>

<h2>BASIC INSPECTION</h2>

```python

print("Rows:", df.shape[0])

print("Columns:", df.shape[1])

df.head()

df.tail()

df.info()

df.columns

df.describe()
```

<h2>CHECK EXPECTED SCHEMA</h2>

```python
expected_columns = [
    "order_id",
    "customer_id",
    "product_id",
    "category",
    "price",
    "discount",
    "quantity",
    "payment_method",
    "order_date",
    "delivery_time_days",
    "region",
    "returned",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age",
    "customer_gender"
]

missing_columns = set(expected_columns) - set(df.columns)

extra_columns = set(df.columns) - set(expected_columns)

print("Missing columns:", missing_columns)
print("Extra columns:", extra_columns)
```

<h2>STANDARDZING THE COLUMNS</h2>

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

<h2>CHECK MISSING VALUES</h2>

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_summary
```

<h2>CHECK DUPLICATE</h2>

```python


duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

#Also check order IDs:

duplicate_orders = df["order_id"].duplicated().sum()

print("Duplicate order IDs:", duplicate_orders)

#If there are duplicates, inspect them:

df[df["order_id"].duplicated(keep=False)].sort_values("order_id")
```

<h2>CHECK DATA TYPES</h2>

```python
df.dtypes

#Convert the date:

df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

#Convert numeric columns:

numeric_columns = [
    "price",
    "discount",
    "quantity",
    "delivery_time_days",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```

```python
df['category'].value_counts()



```

<h2>RETURNED</h2>

```python

df['returned'].value_counts()
```

<h2>STANDARDIZING TEXT</h2>

```python
text_columns = [
    "category",
    "payment_method",
    "region",
    "returned",
    "customer_gender"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

#STANDARDIZE CATEGORY

df["category"] = df["category"].str.title()
```

<h2>BUSINESS RULES</h2>

```python
#PRICE 
df[df["price"] <= 0]

#QUANTITY
df[df['quantity'] <= 0]
```

<h2>DISCOUNTS</h2>

```python
df['discount'].describe()
```

<h2>VALIDATE DATA </h2>

<h2></h2>

```python
#DELIVERY TIME 
df[df["delivery_time_days"] < 0]
# Validate customer age


#The dataset documentation indicates the age range is 18–70.
df[
    (df["customer_age"] < 18) |
    (df["customer_age"] > 70)
]
#Validate return values
valid_returns = ["Yes", "No"]

invalid_returns = df[
    ~df["returned"].isin(valid_returns)
]

invalid_returns
```

<h2>CHECK FINANCIAL CONSISTENCY</h2>

```python
df["calculated_amount"] = (
    df["price"] *
    df["quantity"]
)


df["expected_amount"] = (
    df["price"] *
    df["quantity"] *
    (1 - df["discount"] )
)

df["amount_difference"] = (
    df["total_amount"] -
    df["expected_amount"]
)


df
```
